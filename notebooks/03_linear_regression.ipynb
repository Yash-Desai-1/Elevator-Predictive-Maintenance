```python
import pandas as pd
import matplotlib
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, r2_score, mean_squared_error
import numpy as np
```

```python
df = pd.read_csv('../data/processed/elevator_clean.csv')
```

```python
df.head()
```

```python
df.drop('Hour_Bin', axis=1, inplace=True)
```

```python
df.head()
```

```python
# Features and target
features = ["revolutions", "humidity", "x1", "x2", "x3", "x4", "x5"]
X = df[features]
y = df["vibration"]
```

```python
split = int(0.80 * len(df))
X_train, X_test = X.iloc[:split], X.iloc[split:]
y_train, y_test = y.iloc[:split], y.iloc[split:]
 
```

```python
model = LinearRegression()
model.fit(X_train, y_train)    # ← this is where learning happens

print("  Training complete!")
print()
print("  What the model learned (coefficients):")
for feature, coef in zip(features, model.coef_):
    print(f"    {feature:<14}: {coef:+.4f}")
print(f"    {'intercept':<14}: {model.intercept_:+.4f}")
 
```

```python
y_pred = model.predict(X_test)
```

```python
# now we will calculate the mean absolute error and R^2 score to evaluate the model's performance
mae = mean_absolute_error(y_test, y_pred)
rmse = np.sqrt(mean_squared_error(y_test, y_pred))
r2 = r2_score(y_test, y_pred)

print(f"  Mean Absolute Error: {mae:.4f}")
print(f"  Root Mean Squared Error: {rmse:.4f}")
print(f"  R^2 Score: {r2:.4f}")
```

####
As a baseline, we applied Linear Regression using all 7
  features. Under temporal evaluation, this model achieved
  MAE : 10.97, RMSE : 16.27 and R^2 score : -0.24
  The negative R² indicates the model performs worse than
  simply predicting the mean vibration for every row, confirming
  that the linear relationship between features and vibration
  is insufficient to capture the true dynamics."
