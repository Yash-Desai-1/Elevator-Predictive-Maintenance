```python
split = int(len(df) * 0.80)

y_train = y.iloc[:split]
y_test = y.iloc[split:] # train-test split
```

```python
# Calculate average vibration from training data
mean_value = y_train.mean()

# Predict the same value for every test sample
mean_predictions = np.full(len(y_test), mean_value) # it creates an array of the same length as y_test, filled with the mean_value

# Create some fake good predictions only for comparison
good_predictions = y_test + np.random.normal(0, 5, len(y_test)) # in simple terms, it adds some random noise to the actual test values to simulate a model that predicts well
print(mean_predictions)
print(good_predictions)
```

```python
# then we will calculate MAE, RMSE, and R2 score for both mean_predictions and good_predictions to compare their performance.
errors = abs(y_test - mean_predictions)
mae = np.mean(errors)
print(f'Mean Absolute Error (MAE) for mean predictions: {mae}')
```

```python
# Calculate RMSE for mean predictions
squared_errors = (y_test - mean_predictions) ** 2
rmse = np.sqrt(np.mean(squared_errors))
print(f'Root Mean Squared Error (RMSE) for mean predictions: {rmse}')
```

```python
# R^2 score for mean predictions
ss_total = np.sum((y_test - np.mean(y_test)) ** 2)
ss_residual = np.sum((y_test - mean_predictions) ** 2)
r2 = 1 - (ss_residual / ss_total)
print(f'R^2 Score for mean predictions: {r2}')
```
